Спецификация программы предполагается следующей: есть процедура, реализующая построение границ доверительного интервала для заданной выборки методом бутстреп, условно назовем ее cint = bootci(@est, x). Ей на вход передается указатель или ссылка @est на функцию est, которая реализует вычисление оценки по заданной выборке x: p_est = est(x), где x - вектор элементов выборки, p_est - возвращаемое значение оценки интересующего параметра. Переменная cint представляет собой вектор из двух элементов, являющихся границами доверительного интервала.

```python
import numpy as np
```

```python
def bootci(est, x):
  # Генерируем 1000 бутстреп-выборок с возвращением и вычисляем оценки
  ests = [est(np.random.choice(x, len(x), replace=True)) for _ in range(1000)]
  # Возвращаем границы 95%-го процентильного доверительного интервала
  return np.quantile(ests,[0.025, 0.975])
```

```python
x = np.array([2, 4, 5, 7, 9])

def est(x):
  return np.mean(x)

print(bootci(est, x))
```
