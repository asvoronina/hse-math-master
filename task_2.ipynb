```python
import numpy as np, matplotlib.pyplot as plt

sample = [*map(float, input().split())] #задаем выборку
n = len(sample) #длина выборки
h = float(input()) #ширина окна
x = np.linspace(min(sample), max(sample), 100) #создаем 100 равномерно распределенных точек от минимума до максимума
y = [(1/(n*h))*sum((1/np.sqrt(2*np.pi))*np.exp(-0.5*((x-xi)/h)**2) for xi in sample) for x in x] #берем каждый xi из выборки, смотрим расстояние от x до xi
#складываем вклады всех xi, нормируем

plt.plot(x, y)
plt.show()
```
