```python
from scipy.optimize import minimize
import numpy as np
from scipy.stats import norm
```

```python
def MLE(pdf, x, p0):
  result = minimize(lambda p: #задаем функцию, которую нужно минимизировать
           -np.sum(np.log(pdf(x, p))), p0) #домножаем логарифм на -1 (чтобы максимальное значение превратилось в минимальное)
  return result.x #возвращаем найденный вектор параметров
```

Протестируем на выборке

```python
def pdf(x, p): #обьявляем функцию плотности норм распределения
  return norm.pdf(x, loc=p[0], scale=p[1])
```

```python
np.random.seed(42)
x = np.random.normal(loc=5, scale=2, size=1000) #генерируем выборку
```

```python
p_est = MLE(pdf, x, [4, 3])
print(p_est)
```
