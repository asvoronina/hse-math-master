Cоставить максимально лаконичный код, который рассчитывал бы по заданной выборке следующие точечные оценки:
1) математического ожидания и медианы,
2) дисперсии и среднеквадратического отклонения (два варианта - смещенную и несмещенную оценку при использовании в качестве оценки математического ожидания выборочного среднего),
3) несмещенную оценку коэффициентов асимметрии и эксцесса.

```python
import numpy as np
from scipy.stats import skew, kurtosis
```

```python
x = [2, 4, 100]
mean = np.mean(x)
median = np.median(x)
var_biased = np.var(x, ddof=0)
var_unbiased = np.var(x, ddof=1)
std_biased = np.std(x, ddof=0)
std_unbiased = np.std(x, ddof=1)
skewness = skew(x, bias=False)
excess = kurtosis(x, bias=False)
```

```python
results = {
    "Математическое ожидание": mean,
    "Медиана": median,
    "Смещённая дисперсия": var_biased,
    "Несмещённая дисперсия": var_unbiased,
    "Стандартное отклонение (ddof=0)": std_biased,
    "Стандартное отклонение (ddof=1)": std_unbiased,
    "Асимметрия": skewness,
    "Эксцесс": excess
}

for name, value in results.items():
    print(f"{name}: {value:.3f}")
```
